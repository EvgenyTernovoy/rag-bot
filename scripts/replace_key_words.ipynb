```python
import json
import re
from pathlib import Path

# Пути
MAPPING_FILE = Path("terms_map.json")  # JSON с заменами
INPUT_DIR = Path("output")                        # исходная директория с md-файлами
OUTPUT_DIR = Path("knowledge_base")                      # директория для результатов

# Создаём выходную директорию, если нет
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Загружаем словарь замен
with open(MAPPING_FILE, "r", encoding="utf-8") as f:
    mapping = json.load(f)

# Сортируем ключи по длине (чтобы сначала заменять длинные термины)
sorted_terms = sorted(mapping.keys(), key=lambda x: -len(x))

# Регулярка для поиска терминов
pattern = re.compile(r"\b(" + "|".join(re.escape(term) for term in sorted_terms) + r")\b", flags=re.IGNORECASE)

def replace_terms(match):
    orig = match.group(0)
    # Используем словарь, игнорируя регистр
    replacement = mapping.get(orig, mapping.get(orig.title(), mapping.get(orig.lower(), orig)))
    # Сохраняем первую букву заглавной, если была
    if orig[0].isupper():
        replacement = replacement[0].upper() + replacement[1:]
    return replacement

# Проходим по всем md-файлам в INPUT_DIR
for md_file in INPUT_DIR.glob("*.md"):
    print(f"Обрабатываем {md_file.name} ...")
    text = md_file.read_text(encoding="utf-8")
    new_text = pattern.sub(replace_terms, text)
    # Сохраняем в OUTPUT_DIR с тем же именем
    output_file = OUTPUT_DIR / md_file.name
    output_file.write_text(new_text, encoding="utf-8")
    print(f"✅ Сохранено: {output_file.name}")

print("Все файлы обработаны.")
```

```python
import os
import re
import requests
from bs4 import BeautifulSoup
from pathlib import Path
from tqdm import tqdm

# ==============================
# 🔧 Настройки
# ==============================

# Вставьте сюда URL страниц
URLS = [
    "https://starwars.fandom.com/wiki/R2-D2",
    "https://starwars.fandom.com/wiki/Luke_Skywalker",
    "https://starwars.fandom.com/wiki/Han_Solo",
    "https://starwars.fandom.com/wiki/Leia_Skywalker_Organa_Solo",
    "https://starwars.fandom.com/wiki/Anakin_Skywalker",
    "https://starwars.fandom.com/wiki/Chewbacca",
    "https://starwars.fandom.com/wiki/Obi-Wan_Kenobi",
    "https://starwars.fandom.com/wiki/Landonis_Balthazar_Calrissian",
    "https://starwars.fandom.com/wiki/Darth_Sidious",
    "https://starwars.fandom.com/wiki/Yoda",
    "https://starwars.fandom.com/wiki/C-3PO",
    "https://starwars.fandom.com/wiki/R2-D2",
    "https://starwars.fandom.com/wiki/Wilhuff_Tarkin"
    "https://starwars.fandom.com/wiki/%22Longshot%22_fire-control_system",
"https://starwars.fandom.com/wiki/%22Pheromone_Zone%22_Generator",
"https://starwars.fandom.com/wiki/%22Tulip%22_style_Imperial_work_station",
"https://starwars.fandom.com/wiki/13-BSG-filament_enhancer",
"https://starwars.fandom.com/wiki/343rd_Conference_of_Despotism",
"https://starwars.fandom.com/wiki/473-L_transformer",
"https://starwars.fandom.com/wiki/47_lightsaber_nullifier",
"https://starwars.fandom.com/wiki/A-7_steering_system",
"https://starwars.fandom.com/wiki/A/V_unit",
"https://starwars.fandom.com/wiki/AI_Project",
"https://starwars.fandom.com/wiki/Aaricetri",
"https://starwars.fandom.com/wiki/Abacus",
"https://starwars.fandom.com/wiki/Accelerator_panel",
"https://starwars.fandom.com/wiki/Adventure_room",
"https://starwars.fandom.com/wiki/Adversclick",
"https://starwars.fandom.com/wiki/Advert-sphere",
"https://starwars.fandom.com/wiki/Aeon_Engine",
"https://starwars.fandom.com/wiki/Aerosol_inhaler",
"https://starwars.fandom.com/wiki/Air_barricade",
"https://starwars.fandom.com/wiki/Alarm_system/Legends",
"https://starwars.fandom.com/wiki/Alga-scrubber_convertor_tube",
"https://starwars.fandom.com/wiki/Alsakan_Tessent",
"https://starwars.fandom.com/wiki/Alta-wave_calibration_device",
"https://starwars.fandom.com/wiki/Alta-wave_calibration_device/Legends",
"https://starwars.fandom.com/wiki/Alternating_the_galaxy",
"https://starwars.fandom.com/wiki/Ancient_chalice",
"https://starwars.fandom.com/wiki/Anomaly_1033",
"https://starwars.fandom.com/wiki/Anomaly_2249",
"https://starwars.fandom.com/wiki/Anorelgan_slip-socker",
"https://starwars.fandom.com/wiki/Anti-gravity_water_ball",
"https://starwars.fandom.com/wiki/Aquilae_(The_Star_Wars)",
"https://starwars.fandom.com/wiki/Arachedron_sculpture",
"https://starwars.fandom.com/wiki/Arc_array",
"https://starwars.fandom.com/wiki/Arc_manifold",
"https://starwars.fandom.com/wiki/Archaeotech",
"https://starwars.fandom.com/wiki/Architect%27s_Scepter",
"https://starwars.fandom.com/wiki/Ardent_Crusade",
"https://starwars.fandom.com/wiki/Arges",
"https://starwars.fandom.com/wiki/Arkanian_homeworld",
"https://starwars.fandom.com/wiki/Arkudan_gaming_cube",
"https://starwars.fandom.com/wiki/Armor",
"https://starwars.fandom.com/wiki/Arth%C3%A9zia",
"https://starwars.fandom.com/wiki/Artificial_perspirator",
"https://starwars.fandom.com/wiki/Artificial_weather_system",
"https://starwars.fandom.com/wiki/Assorhan",
"https://starwars.fandom.com/wiki/Atmosphere_containment_projector",
"https://starwars.fandom.com/wiki/Atmospheric_accelerator",
"https://starwars.fandom.com/wiki/Atmospheric_inhibitor",
"https://starwars.fandom.com/wiki/Atomizer_cone",
"https://starwars.fandom.com/wiki/Attractor_node",
"https://starwars.fandom.com/wiki/Aug_scanner",
"https://starwars.fandom.com/wiki/Automatic_winch",
"https://starwars.fandom.com/wiki/Auxiliary_brain",
"https://starwars.fandom.com/wiki/Avi_Singh%27s_vase",
"https://starwars.fandom.com/wiki/Awakening",
"https://starwars.fandom.com/wiki/Ayrou_Tessent",
"https://starwars.fandom.com/wiki/BT-16",
"https://starwars.fandom.com/wiki/Bakuush%27s_centennial_feast",
"https://starwars.fandom.com/wiki/Ballast_tank",
"https://starwars.fandom.com/wiki/Bann_Pu%27dira",
"https://starwars.fandom.com/wiki/Barrier_override",
"https://starwars.fandom.com/wiki/Bartool",
"https://starwars.fandom.com/wiki/Bavana",
"https://starwars.fandom.com/wiki/Beta-wave_mixer",
"https://starwars.fandom.com/wiki/Betting_console",
"https://starwars.fandom.com/wiki/Bherring-24_blinkcode_processing_indicator",
"https://starwars.fandom.com/wiki/Bio-tracker",
"https://starwars.fandom.com/wiki/Biochemical_screen",
"https://starwars.fandom.com/wiki/Biome_cover",
"https://starwars.fandom.com/wiki/Biometrics",
"https://starwars.fandom.com/wiki/Black_Sphere",
"https://starwars.fandom.com/wiki/Blaster_Target_Board",
"https://starwars.fandom.com/wiki/Blaster_wars",
"https://starwars.fandom.com/wiki/Boarding-spike",
"https://starwars.fandom.com/wiki/Braces",
"https://starwars.fandom.com/wiki/Breadboard",
"https://starwars.fandom.com/wiki/Breeze_Globe",
"https://starwars.fandom.com/wiki/Brotherhood_of_the_Ninth_Door_scroll",
"https://starwars.fandom.com/wiki/Burba_Seismic_Activity_Reader",
"https://starwars.fandom.com/wiki/Bursant",
"https://starwars.fandom.com/wiki/CP-0_steering_system",
"https://starwars.fandom.com/wiki/Caisson",
"https://starwars.fandom.com/wiki/Calibrator",
"https://starwars.fandom.com/wiki/Cantonarive_crown_coin",
"https://starwars.fandom.com/wiki/Carbon_scoring",
"https://starwars.fandom.com/wiki/Carbon_scoring/Legends",
"https://starwars.fandom.com/wiki/Catalyst_reactant_cradle",
"https://starwars.fandom.com/wiki/Category_talk:Artifacts?action=edit&redlink=1",
"https://starwars.fandom.com/wiki/Category_talk:Events?action=edit&redlink=1",
"https://starwars.fandom.com/wiki/Category_talk:Planets",
"https://starwars.fandom.com/wiki/Category_talk:Technology",
"https://starwars.fandom.com/wiki/Central_poise_steering_system",
"https://starwars.fandom.com/wiki/Centrifuge",
"https://starwars.fandom.com/wiki/Cerebral_processor",
"https://starwars.fandom.com/wiki/Cerifisis",
"https://starwars.fandom.com/wiki/Ch%27hosk_Cave_Drawings",
"https://starwars.fandom.com/wiki/Chandi_Merle",
"https://starwars.fandom.com/wiki/Charge_looper",
"https://starwars.fandom.com/wiki/Cheater",
"https://starwars.fandom.com/wiki/Chieftain%27s_skull",
"https://starwars.fandom.com/wiki/Chit",
"https://starwars.fandom.com/wiki/Chromium_timelock",
"https://starwars.fandom.com/wiki/Chromomite",
"https://starwars.fandom.com/wiki/Chryselia",
"https://starwars.fandom.com/wiki/Cloaked_smuggling_compartment",
"https://starwars.fandom.com/wiki/Clonetroller",
"https://starwars.fandom.com/wiki/Code_replicator",
"https://starwars.fandom.com/wiki/Coh_Veshiv%27s_homeworld",
"https://starwars.fandom.com/wiki/Cold-weld_arrestor",
"https://starwars.fandom.com/wiki/Colony_marker",
"https://starwars.fandom.com/wiki/Colossus_identification_badge",
"https://starwars.fandom.com/wiki/Compressor_unit",
"https://starwars.fandom.com/wiki/Conduit_purger",
"https://starwars.fandom.com/wiki/Conference_on_Aviles_Prime",
"https://starwars.fandom.com/wiki/Control_box",
"https://starwars.fandom.com/wiki/Cooking_cube",
"https://starwars.fandom.com/wiki/Cooling_unit",
"https://starwars.fandom.com/wiki/Cora_Vessora%27s_homeworld",
"https://starwars.fandom.com/wiki/Cord_fiber",
"https://starwars.fandom.com/wiki/Cornwall",
"https://starwars.fandom.com/wiki/Corrida",
"https://starwars.fandom.com/wiki/Coryio_Chalice",
"https://starwars.fandom.com/wiki/Cosmic_ray_sink",
"https://starwars.fandom.com/wiki/Countermeasures",
"https://starwars.fandom.com/wiki/Creature_Caretakers_Activation_Scan_Point",
"https://starwars.fandom.com/wiki/Credit_changer",
"https://starwars.fandom.com/wiki/Cross-converter",
"https://starwars.fandom.com/wiki/Crystal_growth_tray",
"https://starwars.fandom.com/wiki/Crystal_matrix",
"https://starwars.fandom.com/wiki/Crystal_modulator",
"https://starwars.fandom.com/wiki/Crystal_round",
"https://starwars.fandom.com/wiki/Crystalline_reformatter",
"https://starwars.fandom.com/wiki/Cynestra",
"https://starwars.fandom.com/wiki/D%27Assem",
"https://starwars.fandom.com/wiki/D-93_sonic_tri-denax_cable",
"https://starwars.fandom.com/wiki/Dancing_Goddess",
"https://starwars.fandom.com/wiki/Dancing_Goddess/Legends",
"https://starwars.fandom.com/wiki/DarkStryder_technology",
"https://starwars.fandom.com/wiki/Darth_Vader_figurine",
"https://starwars.fandom.com/wiki/Darvian_Medallion",
"https://starwars.fandom.com/wiki/Dashaad",
"https://starwars.fandom.com/wiki/Data_probe",
"https://starwars.fandom.com/wiki/Deactivator_wand",
"https://starwars.fandom.com/wiki/Deflector_grille",
"https://starwars.fandom.com/wiki/Deflector_shield",
"https://starwars.fandom.com/wiki/Deluxe_Sun_Room",
"https://starwars.fandom.com/wiki/Deprivation_chamber",
"https://starwars.fandom.com/wiki/Desalination_electrode",
"https://starwars.fandom.com/wiki/Desalination_tank",
"https://starwars.fandom.com/wiki/Devaronian_victory_necklace",
"https://starwars.fandom.com/wiki/Dezon%27s_homeworld",
"https://starwars.fandom.com/wiki/Diffuse-Ion_accelerator",
"https://starwars.fandom.com/wiki/Digitizing_pad",
"https://starwars.fandom.com/wiki/Disease_chamber",
"https://starwars.fandom.com/wiki/Disintegration_booth",
"https://starwars.fandom.com/wiki/Disintegrator",
"https://starwars.fandom.com/wiki/Disney_Playground_Planet",
"https://starwars.fandom.com/wiki/Disrupter_box",
"https://starwars.fandom.com/wiki/Docking_clamp",
"https://starwars.fandom.com/wiki/Door_annunciator",
"https://starwars.fandom.com/wiki/Dream-rock",
"https://starwars.fandom.com/wiki/Dream_sucking_device",
"https://starwars.fandom.com/wiki/Droid_deactivator",
"https://starwars.fandom.com/wiki/Droid_vision_phoropter",
"https://starwars.fandom.com/wiki/Duzay",
"https://starwars.fandom.com/wiki/EAU_chip",
"https://starwars.fandom.com/wiki/EP-N5_fastener",
"https://starwars.fandom.com/wiki/Electrochemical_cell",
"https://starwars.fandom.com/wiki/Electrostatic_dome",
"https://starwars.fandom.com/wiki/Elemental_engine",
"https://starwars.fandom.com/wiki/Emergency_button",
"https://starwars.fandom.com/wiki/Emergency_hull_demolition_sequencer",
"https://starwars.fandom.com/wiki/Energy_capsule_(technology)",
"https://starwars.fandom.com/wiki/Epitaxial_discharger",
"https://starwars.fandom.com/wiki/Erla_Tomvaas",
"https://starwars.fandom.com/wiki/Estarvera",
"https://starwars.fandom.com/wiki/Everon",
"https://starwars.fandom.com/wiki/Exigar",
"https://starwars.fandom.com/wiki/Exo_(planet)",
"https://starwars.fandom.com/wiki/Explosive_charge/Legends",
"https://starwars.fandom.com/wiki/Eye_of_the_Albino_Cyclops",
"https://starwars.fandom.com/wiki/Fabled_Key",
"https://starwars.fandom.com/wiki/Fall_of_Yordain_Core",
"https://starwars.fandom.com/wiki/Fault-locator",
"https://starwars.fandom.com/wiki/Fiasco_on_Ryloth",
"https://starwars.fandom.com/wiki/Field_calibrator",
"https://starwars.fandom.com/wiki/Filata",
"https://starwars.fandom.com/wiki/Filtering_unit",
"https://starwars.fandom.com/wiki/Fire_trap",
"https://starwars.fandom.com/wiki/Firebird_of_Colsassa",
"https://starwars.fandom.com/wiki/First_night%27s_reception_dinner_(Chaf_Envoy)",
"https://starwars.fandom.com/wiki/Flash_charge",
"https://starwars.fandom.com/wiki/Flash_shield",
"https://starwars.fandom.com/wiki/Flight_pack",
"https://starwars.fandom.com/wiki/Floppy_Helix_Intensifier",
"https://starwars.fandom.com/wiki/Flotation_collar",
"https://starwars.fandom.com/wiki/Flux_damper",
"https://starwars.fandom.com/wiki/Food_preparation_system",
"https://starwars.fandom.com/wiki/Foran_Tutha_star_probe",
"https://starwars.fandom.com/wiki/Force_bubble",
"https://starwars.fandom.com/wiki/Fractal_Heresy",
"https://starwars.fandom.com/wiki/Fractal_transmitter",
"https://starwars.fandom.com/wiki/Fuel_stabilizer",
"https://starwars.fandom.com/wiki/Fuse_pin",
"https://starwars.fandom.com/wiki/Fusion_bolt",
"https://starwars.fandom.com/wiki/Fusion_portable",
"https://starwars.fandom.com/wiki/Fusion_shell",
"https://starwars.fandom.com/wiki/GCC-1000",
"https://starwars.fandom.com/wiki/Galtor_VI",
"https://starwars.fandom.com/wiki/Gamepack",
"https://starwars.fandom.com/wiki/Garth_Ezzar%27s_homeworld",
"https://starwars.fandom.com/wiki/Gas_Refinery",
"https://starwars.fandom.com/wiki/Gas_mixture_monitor",
"https://starwars.fandom.com/wiki/Gathol",
"https://starwars.fandom.com/wiki/Gausser",
"https://starwars.fandom.com/wiki/Gausser/Legends",
"https://starwars.fandom.com/wiki/Geldoll",
"https://starwars.fandom.com/wiki/Gelemeade",
"https://starwars.fandom.com/wiki/Gem_of_the_Luminous",
"https://starwars.fandom.com/wiki/Gen%27Dai_homeworld",
"https://starwars.fandom.com/wiki/Gen%27Dai_homeworld/Legends",
"https://starwars.fandom.com/wiki/Geological_compressor",
"https://starwars.fandom.com/wiki/Geoscanner",
"https://starwars.fandom.com/wiki/Giddea",
"https://starwars.fandom.com/wiki/Giganism_Sacrifice",
"https://starwars.fandom.com/wiki/Giryulan_Findings",
"https://starwars.fandom.com/wiki/Global_positioning_satellite",
"https://starwars.fandom.com/wiki/Globe_of_Peace",
"https://starwars.fandom.com/wiki/Globe_of_Peace/Legends",
"https://starwars.fandom.com/wiki/Glowbulb",
"https://starwars.fandom.com/wiki/Golden_Agrooba_of_Metlok_IV",
"https://starwars.fandom.com/wiki/Graf_Archive_restoration_effort",
"https://starwars.fandom.com/wiki/Grand_Conflict",
"https://starwars.fandom.com/wiki/Grand_Experiment",
"https://starwars.fandom.com/wiki/Grand_Moot",
"https://starwars.fandom.com/wiki/Grand_Pearl_of_Novak",
"https://starwars.fandom.com/wiki/Granicus",
"https://starwars.fandom.com/wiki/Gravity_seal",
"https://starwars.fandom.com/wiki/Great_Czerka_Datacrash",
"https://starwars.fandom.com/wiki/Great_Fish_Famine",
"https://starwars.fandom.com/wiki/Great_Reconciliation",
"https://starwars.fandom.com/wiki/Great_Schism",
"https://starwars.fandom.com/wiki/Great_Seal",
"https://starwars.fandom.com/wiki/Great_Seal/Legends",
"https://starwars.fandom.com/wiki/Great_Subcrustal_Tubeway",
"https://starwars.fandom.com/wiki/Great_muster",
"https://starwars.fandom.com/wiki/Grimoire_of_Syclos_the_Lame",
"https://starwars.fandom.com/wiki/Groundquake_machine",
"https://starwars.fandom.com/wiki/Guidance_chip",
"https://starwars.fandom.com/wiki/Gung_Slabs",
"https://starwars.fandom.com/wiki/Gusepp%27s_Hand",
"https://starwars.fandom.com/wiki/Han_Solo%27s_dice",
"https://starwars.fandom.com/wiki/Hand_of_Saint_Scorpus",
"https://starwars.fandom.com/wiki/Hand_of_Skrye",
"https://starwars.fandom.com/wiki/Hand_scanner",
"https://starwars.fandom.com/wiki/Handheld_signaler",
"https://starwars.fandom.com/wiki/Harmonic_inducer",
"https://starwars.fandom.com/wiki/Harmonics_counter",
"https://starwars.fandom.com/wiki/Hayananeya",
"https://starwars.fandom.com/wiki/Haysian_ore_medallion",
"https://starwars.fandom.com/wiki/Heating_system",
"https://starwars.fandom.com/wiki/Helicyclic_gear",
"https://starwars.fandom.com/wiki/Hellhulk_station",
"https://starwars.fandom.com/wiki/Herrik",
"https://starwars.fandom.com/wiki/HiSpeed_Blitz_pack",
"https://starwars.fandom.com/wiki/Hieroprince%27s_Ascension",
"https://starwars.fandom.com/wiki/High-resolution_projection_system",
"https://starwars.fandom.com/wiki/Hoe",
"https://starwars.fandom.com/wiki/Holo-News_buyout",
"https://starwars.fandom.com/wiki/Holocron_Machine",
"https://starwars.fandom.com/wiki/Holy_Grail",
"https://starwars.fandom.com/wiki/Home_Nostril_Nap",
"https://starwars.fandom.com/wiki/Hope_of_Glee_Anselm",
"https://starwars.fandom.com/wiki/Hoth_Snow_Machine",
"https://starwars.fandom.com/wiki/Hover_Light",
"https://starwars.fandom.com/wiki/Humidifier",
"https://starwars.fandom.com/wiki/Hydro_transfer_conduit",
"https://starwars.fandom.com/wiki/Hydrodome",
"https://starwars.fandom.com/wiki/Hydrothermal_autoclave",
"https://starwars.fandom.com/wiki/Hypermatter_generator",
"https://starwars.fandom.com/wiki/Hypersonic_oven",
"https://starwars.fandom.com/wiki/Hypno-imprinting",
"https://starwars.fandom.com/wiki/Identi-beacon",
"https://starwars.fandom.com/wiki/Identikit",
"https://starwars.fandom.com/wiki/Impeller_field",
"https://starwars.fandom.com/wiki/Imperial_Customs_Bureau_seal",
"https://starwars.fandom.com/wiki/Imperial_masking_unit",
"https://starwars.fandom.com/wiki/Imperial_transmogrification_unit",
"https://starwars.fandom.com/wiki/Implant_micro-motivator",
"https://starwars.fandom.com/wiki/Inad",
"https://starwars.fandom.com/wiki/Incense_burner/Legends",
"https://starwars.fandom.com/wiki/Indenture_chip",
"https://starwars.fandom.com/wiki/Induction_coil/Legends",
"https://starwars.fandom.com/wiki/Info-scraper",
"https://starwars.fandom.com/wiki/Interphase_module",
"https://starwars.fandom.com/wiki/Ion_burn",
"https://starwars.fandom.com/wiki/Ion_hyper_accelerator",
"https://starwars.fandom.com/wiki/Isolator_aqua-tank",
"https://starwars.fandom.com/wiki/J%27doytzin_III",
"https://starwars.fandom.com/wiki/Jashwa",
"https://starwars.fandom.com/wiki/Jewel_of_Yavin",
"https://starwars.fandom.com/wiki/Jewel_of_Yavin/Legends",
"https://starwars.fandom.com/wiki/Job_on_Niamos",
"https://starwars.fandom.com/wiki/Jod_Na_Nawood%27s_homeworld",
"https://starwars.fandom.com/wiki/Junarian_vase",
"https://starwars.fandom.com/wiki/K-spline",
"https://starwars.fandom.com/wiki/KX-112",
"https://starwars.fandom.com/wiki/Kajim",
"https://starwars.fandom.com/wiki/Kaltrais",
"https://starwars.fandom.com/wiki/Kanger_bowl",
"https://starwars.fandom.com/wiki/Kanz_Disorders",
"https://starwars.fandom.com/wiki/Karltonian_Fiskin",
"https://starwars.fandom.com/wiki/Katiz_Xemir%27s_homeworld",
"https://starwars.fandom.com/wiki/Kaz%27s_lucky_trophy",
"https://starwars.fandom.com/wiki/Key",
"https://starwars.fandom.com/wiki/Key-code_replicator_input/output_jack",
"https://starwars.fandom.com/wiki/Khil_heirloom",
"https://starwars.fandom.com/wiki/Kho_Phon_Farrus%27_machine",
"https://starwars.fandom.com/wiki/Kinetic_blender",
"https://starwars.fandom.com/wiki/Kinzor",
"https://starwars.fandom.com/wiki/Kiss",
"https://starwars.fandom.com/wiki/Kitel_Phard_vase",
"https://starwars.fandom.com/wiki/Klaptun_IV",
"https://starwars.fandom.com/wiki/Koorivar_homeworld",
"https://starwars.fandom.com/wiki/Koorivar_homeworld/Legends",
"https://starwars.fandom.com/wiki/Kooroo_Sundial",
"https://starwars.fandom.com/wiki/Korrilan_IV/Legends",
"https://starwars.fandom.com/wiki/Kyber_mirror",
"https://starwars.fandom.com/wiki/La%27drach",
"https://starwars.fandom.com/wiki/Leox_Gyasi%27s_homeworld",
"https://starwars.fandom.com/wiki/Llewebum",
"https://starwars.fandom.com/wiki/Local_Sitemap",
"https://starwars.fandom.com/wiki/Lodan_job",
"https://starwars.fandom.com/wiki/Logan_Erstar%27s_homeworld",
"https://starwars.fandom.com/wiki/Lop%27s_homeworld",
"https://starwars.fandom.com/wiki/Lost_Tairn",
"https://starwars.fandom.com/wiki/Luminous_Bantha",
"https://starwars.fandom.com/wiki/Maateen",
"https://starwars.fandom.com/wiki/Magic_rock",
"https://starwars.fandom.com/wiki/Magic_wand",
"https://starwars.fandom.com/wiki/Maldran",
"https://starwars.fandom.com/wiki/Malorin",
"https://starwars.fandom.com/wiki/Mangana_aqua_jewel",
"https://starwars.fandom.com/wiki/Marchion_Ro%27s_helmet",
"https://starwars.fandom.com/wiki/Marriten",
"https://starwars.fandom.com/wiki/Masthead_of_Xim_the_Despot",
"https://starwars.fandom.com/wiki/Maz%27s_Curio_Box",
"https://starwars.fandom.com/wiki/Mendacian_funeral_urn",
"https://starwars.fandom.com/wiki/Mhalanduin_Reformation",
"https://starwars.fandom.com/wiki/Mind_Flayer",
"https://starwars.fandom.com/wiki/Mindharp_of_Sharu",
"https://starwars.fandom.com/wiki/Mindharp_of_Sharu/Legends",
"https://starwars.fandom.com/wiki/Mindsplinter",
"https://starwars.fandom.com/wiki/Minstrel",
"https://starwars.fandom.com/wiki/Mission_to_Lupal",
"https://starwars.fandom.com/wiki/Mithra",
"https://starwars.fandom.com/wiki/Monsolar",
"https://starwars.fandom.com/wiki/Morodin_homeworld",
"https://starwars.fandom.com/wiki/Mynock_roast",
"https://starwars.fandom.com/wiki/Nameless_control_rods",
"https://starwars.fandom.com/wiki/Nela-Ray_Resettlement",
"https://starwars.fandom.com/wiki/Nephryn",
"https://starwars.fandom.com/wiki/Nocturne_of_the_Winged-Stars",
"https://starwars.fandom.com/wiki/Nojic_Beanstalk",
"https://starwars.fandom.com/wiki/Norisyn",
"https://starwars.fandom.com/wiki/Ogana_Major",
"https://starwars.fandom.com/wiki/Ohnahmi",
"https://starwars.fandom.com/wiki/Omega%27s_Compass",
"https://starwars.fandom.com/wiki/Onjo_Rhask%27s_Quad_Race_Mayham",
"https://starwars.fandom.com/wiki/Ophuchi",
"https://starwars.fandom.com/wiki/Orb_of_Passage",
"https://starwars.fandom.com/wiki/Ou%27tranoi",
"https://starwars.fandom.com/wiki/Ou%27tranoi/Legends",
"https://starwars.fandom.com/wiki/Packet_of_information",
"https://starwars.fandom.com/wiki/Pall_Gopal%27s_package",
"https://starwars.fandom.com/wiki/Patagonia",
"https://starwars.fandom.com/wiki/Pearl_of_Wisdom",
"https://starwars.fandom.com/wiki/Pedorian",
"https://starwars.fandom.com/wiki/Peekoine",
"https://starwars.fandom.com/wiki/Pelicd",
"https://starwars.fandom.com/wiki/Penegelen_Shards",
"https://starwars.fandom.com/wiki/Perkell_Sector_Mining_Fleecing",
"https://starwars.fandom.com/wiki/Planet",
"https://starwars.fandom.com/wiki/Plaque_of_Victory",
"https://starwars.fandom.com/wiki/Polarca_VII",
"https://starwars.fandom.com/wiki/Power_gem",
"https://starwars.fandom.com/wiki/Proceedings_of_Galactic_Anthropology_and_History",
"https://starwars.fandom.com/wiki/Project_Another_Chance",
"https://starwars.fandom.com/wiki/Promencious_Four",
"https://starwars.fandom.com/wiki/Quermian_holy_relic",
"https://starwars.fandom.com/wiki/Quesoth_pottery",
"https://starwars.fandom.com/wiki/Rahm_Kota%27s_homeworld",
"https://starwars.fandom.com/wiki/Rain_Globe",
"https://starwars.fandom.com/wiki/Rakanal",
"https://starwars.fandom.com/wiki/Rei%27izu",
"https://starwars.fandom.com/wiki/Relics",
"https://starwars.fandom.com/wiki/Remduba_II",
"https://starwars.fandom.com/wiki/Rhydonium_fuel_pipe_siphon",
"https://starwars.fandom.com/wiki/Riah",
"https://starwars.fandom.com/wiki/Riggs_Ashar%27s_homeworld",
"https://starwars.fandom.com/wiki/Ristal",
"https://starwars.fandom.com/wiki/Rod_of_Ages",
"https://starwars.fandom.com/wiki/Rod_of_Daybreak",
"https://starwars.fandom.com/wiki/Rokirex",
"https://starwars.fandom.com/wiki/Romossobar",
"https://starwars.fandom.com/wiki/Roov",
"https://starwars.fandom.com/wiki/Royal_Sovereign_of_Naboo_medal",
"https://starwars.fandom.com/wiki/Rumil",
"https://starwars.fandom.com/wiki/Runaway_Droid_Ride",
"https://starwars.fandom.com/wiki/Rywin",
"https://starwars.fandom.com/wiki/S%C3%A9l%C3%A9ann",
"https://starwars.fandom.com/wiki/Safe_Home",
"https://starwars.fandom.com/wiki/Sagamore",
"https://starwars.fandom.com/wiki/Sand_People_Carving",
"https://starwars.fandom.com/wiki/Sankara_Stones",
"https://starwars.fandom.com/wiki/Sanod",
"https://starwars.fandom.com/wiki/Scouring_of_the_Hamnibus_Glade",
"https://starwars.fandom.com/wiki/Screen_of_the_Second_Sight",
"https://starwars.fandom.com/wiki/Scum_and_Villainy:_Case_Files_on_the_Galaxy%27s_Most_Notorious",
"https://starwars.fandom.com/wiki/Seluket",
"https://starwars.fandom.com/wiki/Selvernis",
"https://starwars.fandom.com/wiki/Senate_Day",
"https://starwars.fandom.com/wiki/Sennoa",
"https://starwars.fandom.com/wiki/Sensor",
"https://starwars.fandom.com/wiki/Shadow_Traproom",
"https://starwars.fandom.com/wiki/Shaw_Uprising",
"https://starwars.fandom.com/wiki/Shingra",
"https://starwars.fandom.com/wiki/Shipwright%27s_Tools",
"https://starwars.fandom.com/wiki/Skeressa",
"https://starwars.fandom.com/wiki/Skull_of_Thalemute",
"https://starwars.fandom.com/wiki/Slumber",
"https://starwars.fandom.com/wiki/Smuggler%27s_Guide",
"https://starwars.fandom.com/wiki/Snee_homeworld",
"https://starwars.fandom.com/wiki/Solodoe_seal",
"https://starwars.fandom.com/wiki/Solstice_Globe",
"https://starwars.fandom.com/wiki/Solstice_V",
"https://starwars.fandom.com/wiki/Soum",
"https://starwars.fandom.com/wiki/Spiral_mist-horn",
"https://starwars.fandom.com/wiki/Staff_of_Command/Legends",
"https://starwars.fandom.com/wiki/Statue_of_Emperor_Palpatine",
"https://starwars.fandom.com/wiki/Sturg_Ganna%27s_homeworld",
"https://starwars.fandom.com/wiki/Subjugation_of_the_False_Apprentice",
"https://starwars.fandom.com/wiki/Suboreen",
"https://starwars.fandom.com/wiki/Sunstar",
"https://starwars.fandom.com/wiki/Survival_Day",
"https://starwars.fandom.com/wiki/Swap_Meet",
"https://starwars.fandom.com/wiki/Syned_One_obelisk",
"https://starwars.fandom.com/wiki/Talman_III",
"https://starwars.fandom.com/wiki/Taris_Affair",
"https://starwars.fandom.com/wiki/Taurill_homeworld",
"https://starwars.fandom.com/wiki/Tech%27s_goggles",
"https://starwars.fandom.com/wiki/Techno-totem",
"https://starwars.fandom.com/wiki/Teklos",
"https://starwars.fandom.com/wiki/Tellion",
"https://starwars.fandom.com/wiki/Temesher",
"https://starwars.fandom.com/wiki/Tensor_4",
"https://starwars.fandom.com/wiki/Tessent",
"https://starwars.fandom.com/wiki/The_Cornerstone_of_the_Galaxy",
"https://starwars.fandom.com/wiki/The_Eye_of_the_Beyonder",
"https://starwars.fandom.com/wiki/The_Golden_Widget",
"https://starwars.fandom.com/wiki/The_Graveyard_of_the_Dragons",
"https://starwars.fandom.com/wiki/The_Package",
"https://starwars.fandom.com/wiki/The_Rebel_Files",
"https://starwars.fandom.com/wiki/The_Tale_of_A_Tiger_and_A_Worm",
"https://starwars.fandom.com/wiki/The_Vanishing",
"https://starwars.fandom.com/wiki/The_galaxy/Legends",
"https://starwars.fandom.com/wiki/Third_Cataclysm",
"https://starwars.fandom.com/wiki/Thornseed",
"https://starwars.fandom.com/wiki/Three_Great_Works",
"https://starwars.fandom.com/wiki/Tinian_Codex",
"https://starwars.fandom.com/wiki/Tiss%27ell",
"https://starwars.fandom.com/wiki/Tormin",
"https://starwars.fandom.com/wiki/Totem_of_Saglanost",
"https://starwars.fandom.com/wiki/Toydarian_clan_bell/Legends",
"https://starwars.fandom.com/wiki/Trancret_Stock_Exchange_Swindle",
"https://starwars.fandom.com/wiki/Trandoshan_relic",
"https://starwars.fandom.com/wiki/Trebold_Raynor%27s_homeworld",
"https://starwars.fandom.com/wiki/Trinity_Stones",
"https://starwars.fandom.com/wiki/Tripartite_Culture_Effrontery",
"https://starwars.fandom.com/wiki/Truant_Sweep",
"https://starwars.fandom.com/wiki/Tyrias",
"https://starwars.fandom.com/wiki/U-One",
"https://starwars.fandom.com/wiki/Ukio_Bread_Incident",
"https://starwars.fandom.com/wiki/Um%27Tal",
"https://starwars.fandom.com/wiki/Umax",
"https://starwars.fandom.com/wiki/Unidentified_26-hour_planet",
"https://starwars.fandom.com/wiki/Unidentified_Corellian_colony",
"https://starwars.fandom.com/wiki/Unidentified_Kwa_artifact",
"https://starwars.fandom.com/wiki/Unidentified_Nikto_colony_world",
"https://starwars.fandom.com/wiki/Unidentified_Tinian_planet",
"https://starwars.fandom.com/wiki/Unidentified_artifact_(Khil)",
"https://starwars.fandom.com/wiki/Unidentified_artifact_(OG-LC)",
"https://starwars.fandom.com/wiki/Unidentified_artifact_(Phee_Genoa)",
"https://starwars.fandom.com/wiki/Unidentified_backwater_planet",
"https://starwars.fandom.com/wiki/Unidentified_cape",
"https://starwars.fandom.com/wiki/Unidentified_chalice",
"https://starwars.fandom.com/wiki/Unidentified_craggy_planet",
"https://starwars.fandom.com/wiki/Unidentified_figurine",
"https://starwars.fandom.com/wiki/Unidentified_gala_fundraiser",
"https://starwars.fandom.com/wiki/Unidentified_green-skied_planet",
"https://starwars.fandom.com/wiki/Unidentified_green_planet",
"https://starwars.fandom.com/wiki/Unidentified_holy_idol",
"https://starwars.fandom.com/wiki/Unidentified_ice_planet_(Hailstorm_Brotherhood)",
"https://starwars.fandom.com/wiki/Unidentified_mask",
"https://starwars.fandom.com/wiki/Unidentified_mountainous_planet_(Spectres)",
"https://starwars.fandom.com/wiki/Unidentified_orange_planet",
"https://starwars.fandom.com/wiki/Unidentified_planet_(A-rack)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Balyeg)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Canderous_Ordo)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Dragon_Void_Run)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Exim_Panshard)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Fermata_Cage)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Gell_Mattar)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Jedi_temple)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Klikklak)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Lupr%27or)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Maltheen_Divide)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Mitaka)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Nes_Ukul)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Nuso_Esva%27s_commanders)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Okinaa)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Outreachial)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Poe_Dameron)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Rhydonium_heist)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Skywalker_and_Tano)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Tholme_and_Ventress)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Unduli_and_Offee)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Vestara_Khai)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Wookiees)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(Younglings)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(blue-skinned_race)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(old_Hamma%27s_place)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(sabacc)",
"https://starwars.fandom.com/wiki/Unidentified_planet_(seven_rings)",
"https://starwars.fandom.com/wiki/Unidentified_red-skied_planet",
"https://starwars.fandom.com/wiki/Unidentified_red_planet",
"https://starwars.fandom.com/wiki/Unidentified_relic",
"https://starwars.fandom.com/wiki/Unidentified_ring",
"https://starwars.fandom.com/wiki/Unidentified_sandy_planet_(Dragon_Void_Run)",
"https://starwars.fandom.com/wiki/Unidentified_tome",
"https://starwars.fandom.com/wiki/Unidentified_uninhabited_refuge_world",
"https://starwars.fandom.com/wiki/Unidentified_urn",
"https://starwars.fandom.com/wiki/Unidentified_world_(Gaze_Electric)",
"https://starwars.fandom.com/wiki/Unidentified_world_(Orla_Jareni)",
"https://starwars.fandom.com/wiki/Unidentified_world_(Voegliss)",
"https://starwars.fandom.com/wiki/Unidentified_yellow_ringed_planet",
"https://starwars.fandom.com/wiki/Unification_Day",
"https://starwars.fandom.com/wiki/Ursellin_Sun_Jammer_Race_Fixings",
"https://starwars.fandom.com/wiki/Useria",
"https://starwars.fandom.com/wiki/Utapau_(Kessil_system)",
"https://starwars.fandom.com/wiki/Vahla_homeworld",
"https://starwars.fandom.com/wiki/Vald",
"https://starwars.fandom.com/wiki/Varovia",
"https://starwars.fandom.com/wiki/Vinzor_Legacy",
"https://starwars.fandom.com/wiki/Volpau%27s_amulet",
"https://starwars.fandom.com/wiki/Vortex_of_Fiery_Dematerialization",
"https://starwars.fandom.com/wiki/Waltar",
"https://starwars.fandom.com/wiki/Weapon",
"https://starwars.fandom.com/wiki/Wind_drum",
"https://starwars.fandom.com/wiki/Wise_Laws_of_Rana",
"https://starwars.fandom.com/wiki/Worrite_homeworld",
"https://starwars.fandom.com/wiki/Wraithwatch",
"https://starwars.fandom.com/wiki/Yallow_soothsaber",
"https://starwars.fandom.com/wiki/Yaloris",
"https://starwars.fandom.com/wiki/Yavin_Vassilika",
"https://starwars.fandom.com/wiki/Yoshiro",
"https://starwars.fandom.com/wiki/Zeitooine_Schism",
"https://starwars.fandom.com/wiki/Zilka_home_planet"
]

# Папка для сохранения результатов
OUTPUT_DIR = Path("output")
OUTPUT_DIR.mkdir(exist_ok=True)

# ==============================
# 🧠 Функции
# ==============================

def clean_text(text: str) -> str:
    """
    Очищает текст для RAG в формате Markdown:
    - убирает сноски и мусор;
    - нормализует переносы строк, пробелы и пунктуацию;
    - сохраняет читаемость и структуру для .md.
    """
    # Убираем сноски [1], [2], ...
    text = re.sub(r"\[\d+\]", "", text)

    # Убираем технические вставки (edit, source, page link и т.п.)
    text = re.sub(r"\([^)]*(edit|source|link|page)[^)]*\)", "", text, flags=re.IGNORECASE)

    # Убираем множественные пробелы и пустые строки
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{2,}", "\n\n", text)

    # Нормализуем пунктуацию
    text = re.sub(r"\s+([,.;:!?])", r"\1", text)

    # Заменяем длинные тире и редкие символы
    text = text.replace("–", "-").replace("—", "-").replace("•", "-")

    return text.strip()

def extract_main_content(soup: BeautifulSoup, entity_name: str) -> str:
    """
    Извлекает контент страницы и формирует Markdown-документ.
    Структура:
    # <entity_name>
    ## <section>
    <text>
    """
    content = soup.find("div", {"class": "mw-parser-output"})
    if not content:
        return f"# {entity_name}\n\n_Нет данных_"

    # Удаляем лишние блоки
    for tag in content.find_all(["table", "style", "script", "figure", "img", "aside", "nav", "noscript", "sup"]):
        tag.decompose()
    for tag in content.find_all(class_=["toc", "disambig", "quote", "reference", "navbox", "mw-editsection"]):
        tag.decompose()

    markdown_lines = [f"# {entity_name}\n"]

    current_section = "Overview"
    current_text = ""

    for el in content.children:
        if el.name in ["h2", "h3"]:
            # Сохраняем предыдущую секцию
            if current_text.strip():
                markdown_lines.append(f"## {current_section}\n\n{clean_text(current_text)}\n")
            # Начинаем новую секцию
            current_section = el.get_text(strip=True)
            current_text = ""
        elif el.name in ["p", "ul", "ol"]:
            current_text += el.get_text(separator=" ", strip=True) + "\n"

    # Добавляем последнюю секцию
    if current_text.strip():
        markdown_lines.append(f"## {current_section}\n\n{clean_text(current_text)}\n")

    markdown_text = "\n".join(markdown_lines).strip()
    return markdown_text

def parse_page(url: str) -> str:
    """
    Загружает страницу и извлекает чистый текст.
    """
    headers = {"User-Agent": "Mozilla/5.0 (compatible; StarWarsParser/1.0)"}
    response = requests.get(url, headers=headers, timeout=15)

    if response.status_code != 200:
        raise Exception(f"Ошибка загрузки {url}: статус {response.status_code}")

    soup = BeautifulSoup(response.text, "html.parser")
    text = extract_main_content(soup)
    return text

def save_text(name: str, text: str):
    """
    Сохраняет текст в файл.
    """
    safe_name = re.sub(r"[^a-zA-Z0-9А-Яа-яёЁ_\-]+", "_", name)
    path = OUTPUT_DIR / f"{safe_name}.txt"
    with open(path, "w", encoding="utf-8") as f:
        f.write(text)
    return path

# ==============================
# 🚀 Основной процесс
# ==============================

for url in tqdm(URLS, desc="Парсинг страниц"):
    try:
        response = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=15)
        soup = BeautifulSoup(response.text, "html.parser")
        entity_name = url.split("/")[-1].replace("_", " ")
        md_text = extract_main_content(soup, entity_name)

        path = OUTPUT_DIR / f"{entity_name}.md"
        with open(path, "w", encoding="utf-8") as f:
            f.write(md_text)

        print(f"✅ Сохранено: {path}")
    except Exception as e:
        print(f"❌ Ошибка при обработке {url}: {e}")
```
